```python
import pandas as pd

df = pd.read_csv("../data/raw/phishing_email.csv")
print(df.shape)
print(df["label"].value_counts())
print(df["text_combined"].isna().sum(), "missing texts")
print(df.duplicated(subset="text_combined").sum(), "duplicate texts")
```

```python
df["length"] = df["text_combined"].astype(str).str.len()
print(df.groupby("label")["length"].describe())
```

```python
for lab in [0, 1]:
    print(f"===== LABEL {lab} =====")
    for t in df[df["label"] == lab]["text_combined"].sample(3, random_state=1):
        print(str(t)[:400], "\n---")
```

```python
from pathlib import Path
import pandas as pd

for f in sorted(Path("../data/raw").glob("*.csv")):
    if f.name == "phishing_email.csv":
        continue
    d = pd.read_csv(f)
    print("=====", f.name, d.shape, d.columns.tolist())
    if "label" in d.columns:
        print(d["label"].value_counts().to_dict())
    print({k: str(v)[:150] for k, v in d.iloc[0].to_dict().items()})
```
